```python
import mysql.connector as connector
```

```python
connection = connector.connect(user = "root", password = "mypass",host="127.0.0.1", db = "littlelemondb")
```

```python
cursor = connection.cursor()
```

```python
show_tables_query = "SHOW tables" 
cursor.execute(show_tables_query)
```

```python
results = cursor.fetchall()
print("Tables in database:")
for table in results:
    print(table[0])
```

```python
promo_query = """
SELECT 
    c.CustomerName,
    c.ContactNumber,
    c.Email,
    o.TotalCost
FROM Customers AS c
INNER JOIN Orders AS o
    ON c.CustomerID = o.Customers_CustomerID
WHERE o.TotalCost > 60;
"""
cursor.execute(promo_query)
```

```python
promo_results = cursor.fetchall()
print("\nCustomers with orders greater than $60:")
for row in promo_results:
    print(f"Name: {row[0]}, Contact: {row[1]}, Email: {row[2]}, Total Cost: ${row[3]}")
```
